```python
import os
import requests
from bs4 import BeautifulSoup
headers = {'Referer':  'www.webtoons.com', 'User-Agent': 'Mozilla/5.0'}
```

```python
#premier URL nécéssaire
res = requests.get("https://www.webtoons.com/...", headers=headers)
soup = BeautifulSoup(res.content, 'html.parser')
```

```python
episode_links = soup.find('div', {"id": "_bottomEpisodeList"}).findAll('a')
# Extraire les URLs des épisodes
episodes = [a.get('href') for a in episode_links if a.get('href')]
```

```python
# Assurez-vous que le chemin vers le dossier de destination soit correcte
from paths import ALL_IMAGES

output_dir = ALL_IMAGES
os.makedirs(output_dir, exist_ok=True)

for y in range(len(episodes)):
    r = requests.get(episodes[y], headers=headers)
    soup = BeautifulSoup(r.content, 'html.parser')
    images = soup.find('div', {"id": "_imageList"}).find_all('img')
    #images = images[1:-1]
    
    for i, img in enumerate(images):
        response = requests.get(img.get('data-url'), headers=headers)
        # Générer un nom de fichier unique
        filename = f"image_ep{y+1}_img{i+1}.png"
        filepath = os.path.join(output_dir, filename)
        with open(filepath, 'wb') as f:
            f.write(response.content)
        print(f"Téléchargé : {filename}")
    print("episode terminé", y,"/",len(episodes))
print("série terminé")
```
